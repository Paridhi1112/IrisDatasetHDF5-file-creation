```python
import h5py
import pandas as pd
import numpy as np
from ucimlrepo import fetch_ucirepo 
import json # For handling metadata dictionaries

print("--- Starting HDF5 Conversion (NumPy 2.0+ compatible) ---")

# 1. Fetch and import iris data 
try:
    print("Fetching Iris dataset from UCImlRepo...")
    # fetch dataset 
    iris = fetch_ucirepo(id=53) 
    
    # data (as pandas dataframes) 
    X = iris.data.features 
    y = iris.data.targets 
    
    print("Dataset fetched successfully.")

    #  2. Convert to HDF5 
    hdf_filename = 'DataScience_Assignment4_Iris_Paridhi_Goyal.hdf5' 
    
    print(f"Creating HDF5 file: {hdf_filename}")

    with h5py.File(hdf_filename, 'w') as f:
        
        # Create the main group for hierarchy
        grp = f.create_group('iris_dataset')

        # --- 3. Save Data (X) and Target (y) ---
        
        # Save features (X)
        data_dataset = grp.create_dataset(
            'data', 
            data=X.values, 
            dtype='float64'
        )
        # Add feature names as an attribute
        feature_names_fixed = [np.bytes_(col) for col in X.columns]
        data_dataset.attrs['feature_names'] = feature_names_fixed
        data_dataset.attrs['units'] = np.bytes_("centimeters (cm)")

        y_fixed_string = y['class'].values.astype(np.bytes_) 
        
        target_dataset = grp.create_dataset(
            'target', 
            data=y_fixed_string
        )
        target_dataset.attrs['column_name'] = np.bytes_(y.columns[0]) 
        target_dataset.attrs['description'] = np.bytes_("Species of Iris flower") 

        #  4. Save Metadata
        
        # == 4a. Save 'iris.metadata' dictionary (simple key-value pairs) ==
        meta_grp = grp.create_group('metadata_uci')
        
        for key, value in iris.metadata.items():
            try:
                if isinstance(value, str):
                    meta_grp.attrs[key] = np.bytes_(value, 'utf-8')
                elif isinstance(value, (int, float, bool)):
                    meta_grp.attrs[key] = value
                elif isinstance(value, list):
                    if all(isinstance(i, str) for i in value):
                         meta_grp.attrs[key] = [np.bytes_(i, 'utf-8') for i in value] 
                    elif all(isinstance(i, (int, float)) for i in value):
                         meta_grp.attrs[key] = np.array(value)
            except TypeError:
                print(f"Warning: Skipping metadata key '{key}' (complex type).")

        # For full preservation, we'll also store the entire dictionary as a JSON string.
        try:
            full_metadata_json = json.dumps(iris.metadata, indent=4)
            meta_grp.attrs['full_metadata_json'] = np.bytes_(full_metadata_json, 'utf-8') 
        except Exception:
            print("Warning: Could not serialize full iris.metadata to JSON.")
            
        # == 4b. Save 'iris.variables' DataFrame (structured metadata) ==
        variables_df = iris.variables.copy()
        for col in variables_df.columns:
            if variables_df[col].dtype == 'object':
                if variables_df[col].apply(lambda x: isinstance(x, list)).any():
                    variables_df[col] = variables_df[col].astype(str)
                
                variables_df[col] = variables_df[col].astype(np.bytes_) # <-- FIX
        
        variables_dataset = grp.create_dataset(
            'variable_information',
            data=variables_df.to_records(index=False)
        )
        variables_dataset.attrs['description'] = np.bytes_("Detailed information on each variable (feature and target) from UCI") # <-- FIX

    print("-" * 30)
    print(f"Successfully created and populated HDF5 file{hdf_filename}")

except Exception as e:
    print(f"An error occurred: {e}")
```
